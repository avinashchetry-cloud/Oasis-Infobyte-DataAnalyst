```python
df_after['Gender'] = df['Gender'].str.strip().str.lower()
def fix_case(row):
    if row['Gender'] == 'female':
        return 'f'
    elif row['Gender'] == 'male':
        return 'm'
    else:
        return row['Gender']

df_after['Gender'] = df_after.apply(fix_case, axis=1)
df_after.head()
```
